```python
import requests
import json

# ใช้ชื่อ Service 'api' แทน localhost เพราะเรายิงหากันภายใน Docker Network
BASE_URL = 'http://api:8000/api/v1'
PDF_PATH = './input/business-manager-elegant-resume-example.pdf' # ให้แก้ชื่อไฟล์ให้ตรงกับที่อยู่ในโฟลเดอร์ input

# 1. Test Extract API
print('--- Testing /extract ---')
with open(PDF_PATH, 'rb') as f:
    res_extract = requests.post(f'{BASE_URL}/extract', files={'file': f})
print(json.dumps(res_extract.json(), indent=2, ensure_ascii=False))
```

```python
# 2. Test Match API
print('\n--- Testing /match ---')
skills = res_extract.json().get('data', {}).get('skill', ['Python', 'Docker', 'MySQL'])
res_match = requests.post(f'{BASE_URL}/match', json={'skills': skills})
if res_match.status_code == 200:
    print(json.dumps(res_match.json(), indent=2, ensure_ascii=False))
else:
    print(f'❌ API Error (Status {res_match.status_code}):', res_match.text)
```

```python
# 3. Test Chat Agent API
print('\n--- Testing /chat ---')
with open(PDF_PATH, 'rb') as f:
    res_chat = requests.post(
        f'{BASE_URL}/chat', 
        files={'file': f}, 
        data={'prompt': 'วิเคราะห์เรซูเม่ หาตำแหน่งงานที่เหมาะสม และสรุปสิ่งที่ควรพัฒนาให้ฟังหน่อย'}
    )
print(res_chat.json().get('answer', 'Error'))
```
