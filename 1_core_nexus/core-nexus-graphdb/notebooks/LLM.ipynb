```python
import time

# ---------------------------------------------------------
# ฟังก์ชันจำลองฝั่ง Decoder (ทายคำถัดไป)
# ---------------------------------------------------------
def mock_llm_predict(decoder_tokens):
    """
    รับรายการ Token ฝั่ง Decoder และประเมินร่วมกับข้อมูลจาก Encoder (จำลอง)
    เพื่อทายคำถัดไป
    """
    # ลำดับคำตอบเป้าหมาย (ไม่รวม [BOS] เพราะ [BOS] คือ Input ตัวแรกสุด)
    target_sequence = [
        "สวัสดี", "ครับ", " ", "ยินดี", "ที่", "ได้", "รู้จัก", "ครับ", " ", 
        "มี", "อะไร", "ให้", "ช่วย", "ไหม", "ครับ", "[EOS]"
    ]
    
    # นับว่าตอนนี้ฝั่ง Decoder มีกี่คำแล้ว (หัก [BOS] ออก 1 ตัว)
    current_step = len(decoder_tokens) - 1 
    
    if current_step < len(target_sequence):
        return target_sequence[current_step]
    else:
        return "[EOS]"

# ---------------------------------------------------------
# โปรแกรมหลัก: จำลอง Dataflow แบบ Encoder-Decoder
# ---------------------------------------------------------

user_input = "สวัสดีครับ ทักทายครับ"

# 1. ฝั่ง ENCODER: ประมวลผลคำถาม (ทำแค่ครั้งเดียว)
# จำลองการตัดคำ (Tokenization) ของ User Input ออกเป็นชิ้นๆ
user_tokens = ["สวัสดี", "ครับ", " ", "ทักทาย", "ครับ"]

# เติม [CLS] ไว้หน้าสุด เพื่อให้ Attention ดึงความหมายรวมของประโยคมาไว้ที่ตัวมัน
encoder_input = ["[CLS]"] + user_tokens

print("=" * 65)
print("🧠 1. ฝั่ง ENCODER (ทำความเข้าใจคำถาม)")
print(f"   📥 Input ที่เข้า Encoder : {encoder_input}")
print("   ... ประมวลผลผ่าน Self-Attention & FFNN ...")
print("   📤 Output               : [Encoded Context Vector (สรุปความหมายการทักทาย)]")
print("=" * 65)
time.sleep(2) # หน่วงเวลาให้เห็นว่าทำงานแยกกัน

# 2. ฝั่ง DECODER: เตรียมจุดสตาร์ท
# เริ่มต้นด้วยสัญลักษณ์ [BOS] เสมอ เพื่อบอกให้เริ่มสร้างคำ
decoder_sequence = ["[BOS]"]

print(f"🚀 2. ฝั่ง DECODER เริ่มทำงาน (สร้างคำตอบทีละคำ)")
print("=" * 65)

loop_count = 1

# 3. เริ่มวนลูป (Autoregressive Loop)
while True:
    time.sleep(1) 
    
    # ส่ง sequence ปัจจุบันไปให้ LLM ทาย (ในของจริงจะมีการดึง Encoded Context มาใช้ใน Cross-Attention ด้วย)
    next_token = mock_llm_predict(decoder_sequence)
    
    print(f"🔄 รอบที่ {loop_count}:")
    print(f"   📥 Decoder Input : {decoder_sequence}")
    print(f"   ✨ ทายคำถัดไป    : '{next_token}'")
    
    if next_token == "[EOS]":
        print(f"\n🛑 เจอ [EOS] -> สิ้นสุดกระบวนการสร้างประโยค!")
        break
        
    decoder_sequence.append(next_token)
    loop_count += 1
    print("-" * 65)

# 4. จัดเรียงผลลัพธ์
# เราจะไม่เอา [BOS] ไปแสดงผลให้ User เห็น
final_response_tokens = decoder_sequence[1:] 
final_text = "".join(final_response_tokens)

print("=" * 65)
print(f"👤 User: {user_input}")
print(f"🤖 LLM : {final_text}")
print("=" * 65)
```
